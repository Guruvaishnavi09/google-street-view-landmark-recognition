```python
import pandas as pd

train_df = pd.read_csv("train.csv")
train_df.shape
```

```python
train_df = pd.read_csv("train.csv", compression="zip")
train_df.shape
```

```python
sample_df = train_df.sample(5000, random_state=42)
sample_df.shape
```

```python
sample_df["landmark_id"].nunique()
```

```python
sample_df["landmark_id"].value_counts().head()
```

```python
top_classes = sample_df["landmark_id"].value_counts().head(20).index

filtered_df = sample_df[sample_df["landmark_id"].isin(top_classes)]

filtered_df.shape
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
filtered_df["label"] = le.fit_transform(filtered_df["landmark_id"])

filtered_df.head()
```

```python
from sklearn.model_selection import train_test_split

train_df, val_df = train_test_split(
    filtered_df,
    test_size=0.2,
    random_state=42,
    stratify=filtered_df["label"]
)

print("Train shape:", train_df.shape)
print("Validation shape:", val_df.shape)
```

```python
filtered_df["image_path"] = "/content/train/" + filtered_df["id"] + ".jpg"
filtered_df.head()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

num_classes = filtered_df["label"].nunique()

model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import numpy as np

# Create dummy image data
X_train = np.random.rand(len(train_df), 224, 224, 3)
X_val = np.random.rand(len(val_df), 224, 224, 3)

y_train = train_df["label"].values
y_val = val_df["label"].values

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=3,
    batch_size=8
)
```

```python
model.evaluate(X_val, y_val)
```
